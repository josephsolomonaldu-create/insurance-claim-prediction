# Insurance Claim Prediction
## 2. Exploratory Data Analysis (EDA)

### Objective
To explore the dataset, identify patterns, relationships and insights that will inform feature engineering and modelling decisions

## 2.1 Load libraries and data
* load necessary libraries
* load raw dataset

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
Train_data = pd.read_csv("Train_data.csv")
```

## 2.2 Target variable Analysis

```python
sns. countplot(x='Claim',
data=Train_data)
plt. title('Distribution of Insurance Claims')
plt.show()
```

## Insights
The distribution of the target variable shows shows a clear class imbalance, with a significantly higher number of buildings that did not record any insurance claim compared to those that did. This imbalance suggests that accuracy alone may be misleading when evaluating model performance. Therefore, metrics such as recall, precision, F1-score and ROC-AUC will be emphasised during model evaluation to better capture the model's ablity to identify buildings with higher claim risk

## 2.3 Univariate Analysis- Numerical features
### Numerical Columns
* Insured_Periods
* Building Dimension
* NumberOfWindows
* YearOfObservations

```python
sns. boxplot(x='Claim',
y='Insured_Period', data=Train_data)
plt.title('Insured Period vs Insurance Claim')
plt.xlabel('Claim')
plt.ylabel('Insured Period')
plt.show()
```

# Insight
Buildings that recoreded insurance claims generally have slightly longer insured periods compared to those without claims. This suggests that longer exposure increases the likelihood of a claim. Although there is some overlap between the two groups, the insured period appears to be relevant predictor and should be retained for modelling.

## 2.4 Univariate Analysis- Categorical features

## Categorical Columns

* Residential
* Building_Painted
* Building_Fenced
* Garden
* Settlement
* Building_Type

```python
sns.countplot(x='Residential',
hue='Claim', data=Train_data)
plt.title('Residential vs Insurance Claim')
plt.show()
```

# EDA insight: Residential vs Insurance Claim
The plot shows that non-residential buildings (Residential = 0) are more frequent in the dataset compared to residential buildings. For residential and non-residential buildings, the number of buildings without claims is significantly higher than those with claims, confirming the class imbalance in the target variable.
However, residential buildings show a relatively higher proportion of insurance claims compared to non-residential buildings. This suggests that residential status may influence the likelihood of filing an insurance claim and should be retained as an important categorical feature for modeling

"Residential is a binary feature where 0 indicates non-residential buildings and 1 indicates residential buildings".

## EDA Summary and Conclusion
Exploratory Data Analysis revealed a class imbalance in the target variable, highlighting the need for appropriate evaluation metrics beyond accuracy. Both numerical and categorical features showed meaningful relationships with insurance claim occurrence. Based on these insights, all relevant features were retained for modeling, with appropriate transformations and encoding applied during preprocessing.
